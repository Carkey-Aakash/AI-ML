# Logistic Regression (Sigmoid + Log-Loss)

## 1. Binary Classification Concept

### What is Binary Classification?
Binary classification is a supervised learning problem where the target variable has **only two classes**.

**Titanic Dataset Example**
- Target variable: `Survived`
  - `1` → Passenger survived
  - `0` → Passenger did not survive

The goal of Logistic Regression is to **predict the probability of survival** for a passenger based on features like:
- Age
- Sex
- Passenger Class (Pclass)
- Fare

---

### How Logistic Regression Works
1. Computes a **linear combination** of input features:
   
   z = w₁x₁ + w₂x₂ + ... + b

2. Applies the **sigmoid function** to convert `z` into a probability.
3. Uses a **threshold (usually 0.5)** to decide the class.

---

### Example
If the model predicts:
- `P(survival) = 0.78` → Passenger **survived**
- `P(survival) = 0.32` → Passenger **did not survive**

---

## 2. Sigmoid Function Behavior

### What is the Sigmoid Function?
The sigmoid function maps any real value into a range **between 0 and 1**, which makes it perfect for probability estimation.

### Mathematical Formula
σ(z) = 1 / (1 + e⁻ᶻ)

---

### Behavior of Sigmoid

| z Value | Sigmoid Output | Meaning |
|-------|---------------|--------|
| Very large +ve | ≈ 1 | Very high survival chance |
| 0 | 0.5 | Uncertain |
| Very large −ve | ≈ 0 | Very low survival chance |

---

### Titanic-Based Intuition
- Female passenger, 1st class → **z = +3.2** → sigmoid ≈ **0.96**
- Male passenger, 3rd class → **z = -2.1** → sigmoid ≈ **0.11**

---

### Why Sigmoid is Used
- Output can be interpreted as **probability**
- Smooth and differentiable → allows **gradient descent**
- Ideal for **binary outcomes**

---

### Limitation
- Saturates at 0 and 1 → gradients become very small  
  (known as **vanishing gradient**)

---

## 3. Log-Loss (Binary Cross-Entropy)

### What is Log-Loss?
Log-loss measures **how far the predicted probability is from the actual label**.

It penalizes **confident but wrong predictions heavily**.

---

### Formula
Loss = − [ y log(p) + (1 − y) log(1 − p) ]

Where:
- y = true label (0 or 1)
- p = predicted probability

---

### Titanic Example

#### Case 1 (Good Prediction)
- Actual: Survived (1)
- Predicted: 0.92  
- Loss → very small

#### Case 2 (Bad Prediction)
- Actual: Survived (1)
- Predicted: 0.05  
- Loss → very high

---

### Why Not Mean Squared Error (MSE)?
- MSE gives weak gradients for wrong predictions
- Log-loss strongly corrects **overconfident mistakes**
- Log-loss is **convex** for logistic regression → faster convergence

---

### Loss Behavior Summary

| Prediction | Actual | Loss |
|----------|--------|------|
| 0.99 | 1 | Very low |
| 0.50 | 1 | Medium |
| 0.01 | 1 | Very high |

---

## 4. How Everything Connects

1. Logistic Regression computes a linear score (z)
2. Sigmoid converts score into probability
3. Log-loss evaluates prediction quality
4. Gradient Descent updates weights to minimize log-loss

---

## 5. Key Points 

- Logistic Regression is **not actually regression**, it is a **classification algorithm**
- Output is **probability**, not class
- Decision boundary is **linear**
- Uses **sigmoid + log-loss**
- Best for **binary classification problems** like Titanic survival

---

## 6. One-Line  Answers

- Logistic Regression predicts probability using sigmoid.
- Sigmoid squashes values between 0 and 1.
- Log-loss penalizes confident wrong predictions.
- Threshold converts probability into class.

### 24. Train the model for 100 epochs and store loss values.

```python
import math
def sigmoid(z):
    return (1/(1+ math.exp(-z)))

def predict_proba(X,w,b):
    probabilities = []
    for x in X:
        z = w * x + b
        p = sigmoid(z)
        probabilities.append(p)
    return probabilities

def log_loss(y_true, y_pred):
    n = len(y_true)
    loss= 0.0
    eps = 1e-15     # to avoid log(0)
    for i in range(n):
        y=y_true[i]
        p=y_pred[i]
        p = max(eps, min(1-eps,p))
        loss+= y * math.log(p) + (1-y) * math.log(1-p)
    return -loss / n


def compute_gradient_full(X, y, w, b):
    n = len(X)
    dw = 0.0
    db = 0.0
    
    for i in range(n):
        x_i = X[i]
        y_i = y[i]
        
        # predicted probability
        y_hat = 1 / (1 + math.exp(-(w*x_i + b)))
        
        # individual gradients for this sample
        dw += (y_hat - y_i) * x_i
        db += (y_hat - y_i)
        
    
    
    # take average
    dw /= n
    db /= n
    
    return dw, db

# Initialize
w = 0.0
b = 0.0
alpha = 0.0001
epochs = 100

losses = []  # store loss values

for epoch in range(epochs):
    # Step 1: Predict
    y_pred = predict_proba(X, w, b)
    
    # Step 2: Compute loss
    loss = log_loss(y, y_pred)
    losses.append(loss)
    
    # Step 3: Compute gradients
    dw, db = compute_gradient_full(X, y, w, b)
    
    # Step 4: Update parameters
    w = w - alpha * dw
    b = b - alpha * db

print("Final w:", w)
print("Final b:", b)
print("First 5 losses:", losses[:5])
print("Last 5 losses:", losses[-5:])
```

### 27. Write code to count false positives and false negatives.

False Positive (FP): Model predicts 1 (Survived) but actual is 0

False Negative (FN): Model predicts 0 (Did not survive) but actual is 1

```python
import math
def sigmoid(z):
    return 1/ (1+math.exp(-z))

def predict_proba(X,w,b):
    probabilities = []
    for x in X:
        z = w *x +b
        p =sigmoid(z)
        probabilities.append(p)
    return probabilities

def compare_predictions(probabilities,threshold=0.7):
    predictions = []
    for probability in probabilities:
        if probability >= threshold:
            predictions.append(1)
        else:
            predictions.append(0)
    return predictions

        
def log_loss(y_true, y_pred):
    n = len(y_true)
    loss=0.0
    eps = 1e-15  # to avoid log(0)
    for i in range(n):
        y = y_true[i]
        p = y_pred[i]
        p=max(eps, min(1-eps,p))
        loss+= y*math.log(p) + (1-y) * math.log(1-p)
    return -loss/n

def compute_gradient_full(X,y,w,b):
    n = len(X)
    dw = 0
    db = 0

    for i in range(n):
        x_i = X[i]
        y_i = y[i]
        y_hat = 1/(1+math.exp(-(w*x_i + b)))
        dw+= (y_hat - y_i) * x_i
        db += (y_hat - y_i)

    dw = dw / n 
    db = db / n

    return dw ,db


def train_logistic_regression(X, y, epochs=100, alpha=0.0001):
    w = 0.0
    b = 0.0
    losses = []
    
    for epoch in range(epochs):
        y_prob = predict_proba(X, w, b)
        loss = log_loss(y, y_prob)
        losses.append(loss)
        
        dw, db = compute_gradient_full(X, y, w, b)
        w -= alpha * dw
        b -= alpha * db
    
    return w, b, losses

def count_number(y_true,y_prediction):
    fp = 0  # false positives
    fn = 0  # false negatives
    for x,y in zip(y_true,y_prediction):
        if x == 0 and y == 1:
            fp+=1
        elif x == 1 and y==0 :
            fn +=1
    return fp, fn

X = data["Fare"]
y = data["Survived"]
# lets fisrt train our logistic regression
w,b,losses = train_logistic_regression(X,y,epochs =100, alpha =0.0001)
print("Final w: ",w)
print("Final b: ",b)

# lets use this final w,b for predictions
pred1 = predict_proba(X,w,b)

predic = compare_predictions(pred1,threshold = 0.7)

fp, fn  = count_number(y, predic)

print("The total number of false positive at threshold 0.7 is",fp)
print("The total number of false negative at threshold 0.7 is",fn)

predic1 = compare_predictions(pred1,threshold = 0.5)
fp1, fn1  = count_number(y, predic1)
print("The total number of false positive at threshold 0.5 is",fp1)
print("The total number of false negative at threshold 0.5 is",fn1)
```

### 29 Modify the model to support multiple input features.

```python
import csv
import math

# -----------------------------
# Step 1: Load dataset safely
# -----------------------------
def load_multifeature_dataset(filepath):
    X = []
    y = []
    with open(filepath, "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            try:
                fare = float(row["Fare"])
                pclass = float(row["Pclass"])
                survived = int(row["Survived"])
                # Only include if none of the features are missing
                X.append([fare, pclass])
                y.append(survived)
            except:
                continue  # skip rows with missing/invalid data
    return X, y

# -----------------------------
# Step 2: Sigmoid function
# -----------------------------
def sigmoid(z):
    return 1 / (1 + math.exp(-z))

# -----------------------------
# Step 3: Predict probabilities
# -----------------------------
def predict_proba_multi(X, w, b):
    probabilities = []
    for x_vector in X:
        z = b
        for i in range(len(w)):
            z += w[i] * x_vector[i]
        probabilities.append(sigmoid(z))
    return probabilities

# -----------------------------
# Step 4: Predict class labels
# -----------------------------
def predict_class(probabilities, threshold=0.5):
    return [1 if p >= threshold else 0 for p in probabilities]

# -----------------------------
# Step 5: Log-loss function
# -----------------------------
def log_loss(y_true, y_pred):
    n = len(y_true)
    loss = 0.0
    eps = 1e-15
    for yi, pi in zip(y_true, y_pred):
        pi = max(eps, min(1 - eps, pi))
        loss += yi * math.log(pi) + (1 - yi) * math.log(1 - pi)
    return -loss / n

# -----------------------------
# Step 6: Compute gradients
# -----------------------------
def compute_gradients_multi(X, y, w, b):
    n = len(X)
    dw = [0.0] * len(w)
    db = 0.0
    
    for xi, yi in zip(X, y):
        z = b
        for j in range(len(w)):
            z += w[j] * xi[j]
        y_hat = sigmoid(z)
        error = y_hat - yi
        for j in range(len(w)):
            dw[j] += error * xi[j]
        db += error
    
    # Average gradients
    dw = [d / n for d in dw]
    db /= n
    
    return dw, db

# -----------------------------
# Step 7: Train logistic regression
# -----------------------------
def train_logistic_regression_multi(X, y, epochs=100, alpha=0.0001):
    n_features = len(X[0])
    w = [0.0] * n_features
    b = 0.0
    losses = []
    
    for epoch in range(epochs):
        y_prob = predict_proba_multi(X, w, b)
        loss = log_loss(y, y_prob)
        losses.append(loss)
        
        dw, db = compute_gradients_multi(X, y, w, b)
        for j in range(n_features):
            w[j] -= alpha * dw[j]
        b -= alpha * db
    
    return w, b, losses

# -----------------------------
# Step 8: Count false positives/negatives
# -----------------------------
def count_fp_fn(y_true, y_pred):
    fp = 0
    fn = 0
    for actual, pred in zip(y_true, y_pred):
        if actual == 0 and pred == 1:
            fp += 1
        elif actual == 1 and pred == 0:
            fn += 1
    return fp, fn

# -----------------------------
# Step 9: Main execution
# -----------------------------
X, y = load_multifeature_dataset("Titanic-Dataset.csv")

# Train the model
w, b, losses = train_logistic_regression_multi(X, y, epochs=100, alpha=0.0001)
print("Final weights:", w)
print("Final bias:", b)
print("Final loss:", losses[-1])

# Predict probabilities and classes
y_prob = predict_proba_multi(X, w, b)
y_pred_05 = predict_class(y_prob, threshold=0.5)
y_pred_07 = predict_class(y_prob, threshold=0.7)

# Count false positives and false negatives
fp_05, fn_05 = count_fp_fn(y, y_pred_05)
fp_07, fn_07 = count_fp_fn(y, y_pred_07)

print("\nThreshold 0.5 → False Positives:", fp_05, "False Negatives:", fn_05)
print("Threshold 0.7 → False Positives:", fp_07, "False Negatives:", fn_07)
```
