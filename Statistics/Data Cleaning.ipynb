## 1. Data Cleaning Basics

**Definition:**  
Data cleaning is the process of fixing or removing incorrect, corrupted, incomplete, duplicated, or improperly formatted data.

### Why Data Cleaning Is Important?
- Improves accuracy of analysis  
- Reduces noise in data  
- Helps machine learning models perform better  
- Avoids misleading insights  

### Common Data Cleaning Steps

1. **Handling Missing Values**
   - Remove missing rows
   - Fill missing values with mean/median/mode
   - Forward/Backward fill
   - Use interpolation

2. **Removing Duplicates**
   - Identify duplicate rows  
   - Drop duplicates

3. **Fixing Data Types**
   - Convert strings to integers/floats  
   - Convert date columns to `datetime`  

4. **Handling Outliers**
   - Use IQR method  
   - Use Z-score  
   - Cap/floor extreme values  

5. **Standardizing & Normalizing**
   - Scaling numeric values  
   - Converting values to the same format  

6. **Cleaning Text Data**
   - Remove punctuation  
   - Remove extra spaces  
   - Convert to lowercase  
   - Remove stopwords  

---

## 2. Regex for Text Cleaning

**Definition:**  
Regular Expressions (Regex) are patterns used to match, search, and clean text.

### Why Use Regex?
- Removes unwanted text  
- Extracts patterns  
- Cleans messy strings quickly  
- Validates formats (email, phone numbers, etc.)  

### Common Regex Patterns for Cleaning

| Task | Pattern | Example |
|------|---------|---------|
| Remove digits | `\d+` | `"abc123" → "abc"` |
| Remove non-alphabetic | `[^a-zA-Z]` | `"Hello123!" → "Hello"` |
| Remove multiple spaces | `\s+` | `"Hello    World" → "Hello World"` |
| Get only numbers | `\d+` | `"Age: 25" → 25` |
| Get only alphabets | `[A-Za-z]+` | `"Room20A" → "Room"` |
| Replace special characters | `[^a-zA-Z0-9 ]` | `"hello@#!" → "hello"` |

### Python Examples

```python
import re

text = "Hello!!! I am 20 years old...   "

# Remove special characters
clean1 = re.sub(r'[^a-zA-Z0-9 ]', '', text)

# Remove extra spaces
clean2 = re.sub(r'\s+', ' ', text).strip()

# Remove numbers
clean3 = re.sub(r'\d+', '', text)

print(clean1)
print(clean2)
print(clean3)

**1. Write a clean_price() function that uses regex to remove currency symbols (₹, $, £) and convert the result into a float.**

````python
import re
def clean_price(price):
    clean = re.sub(r'[₹$£,]','', price)
    clean = clean.strip()
    return float(clean)

print(clean_price("₹500"))
print(clean_price("$1,500"))
print(clean_price("£5,500"))
````

**2. Write a clean_rating() function that extracts only the numeric rating from text like “Rated 4 out of 5”.**

````python
import re 
def clean_rating(text):
    match = re.search(r'\d+(\.\d+)?' ,text)
    if match:
        return float(match.group())
    return None
        
print(clean_rating("4 out of 5 "))
print(clean_rating("Rating: 4.5/5"))        
print(clean_rating("4 stars"))              
print(clean_rating("Rated: 3")) 
````

**3. Write a clean_text() function that removes hashtags, emojis, special symbols, and extra spaces using regex substitutions.**

````python
import re

def clean_text(text):
   
    if not text:
        return ""
    #Remove Hashtags
    clean1 = re.sub(r'#\S+','',text)
    # Remove unwanted characters (everything except letters, numbers, space)
    clean = re.sub(r'[^a-zA-Z0-9 ]', '', clean1)
    
    # Replace multiple spaces/newlines with a single space
    clean = re.sub(r'\s+', ' ', clean).strip()
    
    return clean

# Example usage: messy, real-world text
sample_text = """
Wow!!! This product   is amazing 😍💯
## #Price is $499 only, totally worth it!!!
#BestPurchase #HappyCustomer
Visit www.example.com now!!!
"""

cleaned_text = clean_text(sample_text)
print("Original Text:\n", sample_text)
print("\nCleaned Text:\n", cleaned_text)
````

**4. Given a list of raw scraped items (title, price, rating), write a function clean_item_dict() that returns a cleaned dictionary for each item.**

````python
import re

# --- Cleaning functions ---

def clean_price(price):
    """Remove currency symbols and convert to float"""
    clean = re.sub(r'[₹$£,]', '', price).strip()
    return float(clean) if clean else 0.0

def clean_rating(text):
    """Extract first number as rating"""
    match = re.search(r'\d+(\.\d+)?', text)
    return float(match.group()) if match else 0.0

def clean_text(text):
    """Remove special symbols, emojis, extra spaces"""
    if not text:
        return ""
    clean = re.sub(r'[^a-zA-Z0-9 ]', '', text)
    clean = re.sub(r'\s+', ' ', clean).strip()
    return clean

# --- Function to clean a full item dictionary ---
def clean_item_dict(item):
    """
    Input: raw dictionary with keys 'title', 'price', 'rating'
    Output: cleaned dictionary
    """
    cleaned_item = {}
    
    # Clean each field
    cleaned_item['title'] = clean_text(item.get('title', ''))
    cleaned_item['price'] = clean_price(item.get('price', '0'))
    cleaned_item['rating'] = clean_rating(item.get('rating', '0'))
    
    return cleaned_item

# --- Example usage ---
raw_items = [
    {"title": "Amazing Product!!! 😍", "price": "₹1,299", "rating": "Rated 4.5 out of 5"},
    {"title": "Cheap & Good$ Item", "price": "$499", "rating": "3/5"},
    {"title": "Luxury Item $$$", "price": "£2,000", "rating": "Rated: 5"}
]

cleaned_items = [clean_item_dict(item) for item in raw_items]

for item in cleaned_items:
    print(item)
````

**5. Apply all cleaning functions to a list of items and store the cleaned output in a list called cleaned_data.**

````python
import re

# --- Cleaning functions ---
def clean_price(price):
    return float(re.sub(r'[₹$£,]', '', price).strip())

def clean_rating(text):
    match = re.search(r'\d+(\.\d+)?', text)
    return float(match.group()) if match else 0.0

def clean_text(text):
    clean = re.sub(r'[^a-zA-Z0-9 ]', '', text)
    return re.sub(r'\s+', ' ', clean).strip()

# --- Function to clean a single item ---
def clean_item_dict(item):
    return {
        'title': clean_text(item.get('title', '')),
        'price': clean_price(item.get('price', '0')),
        'rating': clean_rating(item.get('rating', '0'))
    }

# --- Raw items list ---
raw_items = [
    {"title": "Amazing Product!!! 😍", "price": "₹1,299", "rating": "Rated 4.5 out of 5"},
    {"title": "Cheap & Good$ Item", "price": "$499", "rating": "3/5"},
    {"title": "Luxury Item $$$", "price": "£2,000", "rating": "Rated: 5"}
]

# --- Apply cleaning and store in a list called cleaned_data ---
cleaned_data = []
for item in raw_items:
    cleaned_data.append(clean_item_dict(item))

# --- Print cleaned_data ---
print(cleaned_data)
````

**6. Create a validation step: check if cleaned price is numeric, rating is between 1–5, and text is non-empty. If not valid, store the item in a list called rejected_items.**

````python
rejected_items = []

def validate_item(price, rating, text):
    # Check if price is numeric
    if not str(price).isdigit():
        rejected_items.append((price, rating, text))
        return False

    # Check if rating is between 1–5
    if rating < 1 or rating > 5:
        rejected_items.append((price, rating, text))
        return False

    # Check if text is non-empty
    if text.strip() == "":
        rejected_items.append((price, rating, text))
        return False

    return True

# Example 

items = [
    ("100", 5, "Nice product"),
    ("abc", 4, "Good"),       # invalid price
    ("250", 7, "Super"),      # invalid rating
    ("300", 3, "   "),        # empty text
]

for price, rating, text in items:
    if validate_item(price, rating, text):
        print("VALID:", price, rating, text)
    else:
        print("REJECTED:", price, rating, text)

print("\nRejected Items:", rejected_items)
````
