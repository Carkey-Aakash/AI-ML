# Descriptive Statistics & Distribution Types

## 1. Descriptive Statistics

Descriptive statistics are used to **summarize and describe the main features of a dataset**. They help us understand the data before performing any advanced analysis or building models.

### 1.1 Measures of Central Tendency

These measures describe the **center** of the data.

* **Mean**: The average value
  Example:
  Data = [10, 20, 30]
  Mean = (10 + 20 + 30) / 3 = 20

* **Median**: The middle value when data is sorted
  Example:
  Data = [10, 20, 30]
  Median = 20

* **Mode**: The most frequently occurring value
  Example:
  Data = [1, 2, 2, 3]
  Mode = 2

---

### 1.2 Measures of Dispersion

These measures describe **how spread out the data is**.

* **Range**: Difference between maximum and minimum values
  Example:
  Data = [5, 10, 15]
  Range = 15 − 5 = 10

* **Variance**: Average of squared differences from the mean
  It shows how far values are from the mean.

* **Standard Deviation**: Square root of variance
  It tells how much data typically deviates from the mean.

---

### 1.3 Measures of Shape

These measures describe the **shape of the data distribution**.

* **Skewness**: Indicates data asymmetry

  * Positive skew → tail on the right
  * Negative skew → tail on the left
  * Zero skew → symmetric

* **Kurtosis**: Measures how peaked or flat the distribution is

  * High kurtosis → sharp peak
  * Low kurtosis → flat peak

---

## 2. Distribution Types

A distribution shows **how data values are spread across different values**.

---

### 2.1 Normal Distribution

* Symmetric and bell-shaped
* Mean = Median = Mode
* Most values are near the center

Example:
Human height, exam scores

---

### 2.2 Positively Skewed Distribution (Right-Skewed)

* Long tail on the right side
* Mean > Median

Example:
Income distribution, ticket prices

---

### 2.3 Negatively Skewed Distribution (Left-Skewed)

* Long tail on the left side
* Mean < Median

Example:
Easy exam scores where most students score high

---

### 2.4 Uniform Distribution

* All values occur with equal frequency

Example:
Rolling a fair die

---

### 2.5 Bimodal Distribution

* Two peaks in the data

Example:
Heights of male and female groups combined

---

## 3. Why These Concepts Are Important

* Help understand data before modeling
* Identify outliers and data issues
* Choose the right preprocessing techniques
* Improve model accuracy and reliability

---

## 4. Simple Summary

* Descriptive statistics explain **what the data looks like**
* Distributions explain **how data is spread**
* Clean and well-understood data leads to better predictions



**1. Implement the following functions without using NumPy or Pandas built-in statistical functions:**
```
• mean(data)
• median(data)
• mode(data)
• variance(data)
• std(data)



````python
def mean(data):
    if not data:
        return None
    return sum(data) / len(data)
    
def median(data):
    if not data:
        return None
         
    data = sorted(data)
    n = len(data)
    mid = n // 2
    if n % 2 == 0:
        return (data[mid-1]+data[mid])/ 2
    else:
        return data[mid]

def mode(data):
    if not data:
        return None
    freq = {}
    for num in data:
        freq[num] = freq.get(num, 0) + 1

    max_freq = max(freq.values())
    modes = [k for k, v in freq.items() if v == max_freq]
    return modes

def variance(data):
    if not data:
        return None
    m = mean(data)
    return sum((x-m)**2 for x in data) / len(data)

def std_dev(data):
    if not data:
        return None
    var = variance(data)
    return var ** 0.5

        
        
````

````python
import csv

# Load Titanic dataset numerical columns (Age and Fare)
def load_titanic_numerical(file_path):
    numerical_data = {'Age': [], 'Fare': []}
    with open(file_path, "r", newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)

        for row in reader:
            age_val = row["Age"]
            fare_val =row["Fare"]
            numerical_data["Age"].append(float(age_val) if age_val != '' else None)
            numerical_data["Fare"].append(float(fare_val) if fare_val != '' else None)
        return numerical_data
````

**6. After removing outliers (using your Day-1 outlier function), recompute mean and standard deviation
and compare the results with the original values.**

````python
def get_quartile(data, q):
    
    if not data:
        return None
    if q not in (1, 2, 3):
        raise ValueError("q must be 1, 2, or 3")
         
    data = sorted(data)
    n = len(data)
    pos = q * (n+1) / 4

    if pos<=1:
        return data[0]
    if pos>=n:
        return data[-1]
        
    lower = int(pos) - 1
    upper = lower + 1
    fraction = pos - lower
    return data[lower] + fraction * (data[upper] - data[lower])


def get_outliers(data , method = "iqr"):
    outliers = []
    clean = [x for x in data if x is not None]
    if method == "iqr":
        q3 = get_quartile(clean , 3)
        q1 = get_quartile(clean, 1)
        iqr = q3 - q1 
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        for num in clean:
            if num < lower or num > upper:
                outliers.append(num)
    elif method == "zscore":
        m = mean(clean)
        sd = std_dev(clean)
        for x in clean:
            z =abs((x-m) / sd)
            if z > 3:
                outliers.append(x)
    else:
        raise ValueError("Method can only be iqr and zscore")

    return outliers

outlier_iqr = get_outliers(cleaned, method="iqr")
outlier_z = get_outliers(cleaned, method="zscore")

print(f"\nThe outliers using IQR : {len(outlier_iqr)}")    
print(f"\nThe outliers using IQR : {(outlier_iqr)}")  
print(f"\nThe outliers using Z-score : {len(outlier_z)}")  
print(f"\nThe outliers using Z-score : {(outlier_z)}")  

# compare using IQR method
no_outliers_data_iqr = [x for x in cleaned if x not in outlier_iqr]    # data with no outliers
print(f"\nBefore outliers removal using IQR : Mean={mean(cleaned)}, Std_dev={std_dev(cleaned)}")
print(f"\nAfter  outliers removal using IQR : Mean={mean(no_outliers_data_iqr)}, Std_dev={std_dev(no_outliers_data_iqr)}")

# compare using z-score method
no_outliers_data_z = [x for x in cleaned if x not in outlier_z]    # data with no outliers
print(f"\nBefore outliers removal using Zscore: Mean={mean(cleaned)}, Std_dev={std_dev(cleaned)}")
print(f"\nAfter  outliers removal using Zscore: Mean={mean(no_outliers_data_z)}, Std_dev={std_dev(no_outliers_data_z)}")
````

**7. Divide the data into bins (class intervals) and create a simple frequency table using plain Python (no Pandas groupby).**

````python
bin_width = 4
min_age = int(min(cleaned))
# print(min_age)
max_age = int(max(cleaned))

print("\nAge Frequency Table (continuous bins):")

start = min_age
while start <= max_age:
    end = start + bin_width
    count = 0
    for age in cleaned:
        if start <= age < end:   # include start, exclude end for continuous bins
            count += 1
    print(f"{start}-{end}: {count}")
    start = end  # next bin starts at previous end
````
