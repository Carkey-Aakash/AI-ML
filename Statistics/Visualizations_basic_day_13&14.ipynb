# Visualization Basics: Histograms & Boxplots

Data visualization helps us quickly understand data patterns, distribution, and outliers.

---

## Histogram

A **histogram** shows the **distribution of numerical data** using bars.

**Key points:**

* X-axis → Data ranges (bins)
* Y-axis → Frequency
* Shows data distribution and skewness

**Helps identify:**

* Normal, left-skewed, or right-skewed data
* Spread and concentration of values

**Best used for:** Understanding data distribution

---

## Boxplot (Box-and-Whisker Plot)

A **boxplot** summarizes data using the **five-number summary**:

* Minimum, Q1, Median, Q3, Maximum

**Key points:**

* Box → IQR (Q3 − Q1)
* Line inside box → Median
* Points → Outliers

**Helps identify:**

* Median and spread
* Outliers
* Comparison between datasets

**Best used for:** Outlier detection and comparison

---

## Histogram vs Boxplot

| Feature            | Histogram | Boxplot |
| ------------------ | --------- | ------- |
| Distribution shape | Yes       | No      |
| Outliers           | No        | Yes     |
| Comparison         | Moderate  | Easy    |

---

**Conclusion:**
Histograms show how data is distributed, while boxplots summarize data and highlight outliers.

**5. Create a boxplot for a numerical column and identify the lower and upper quartiles.**

```python
import matplotlib.pyplot as plt

# Clean the data first
clean_mdn_house = [x for x in mdn_house if x is not None]

def get_quartile(data, q):
    if not data:
        return None
    if q not in (1, 2, 3):
        raise ValueError("q must be 1, 2, or 3")
         
    data = sorted(data)
    n = len(data)
    pos = q * (n+1) / 4

    if pos <= 1:
        return data[0]
    if pos >= n:
        return data[-1]
        
    lower = int(pos) - 1
    upper = lower + 1
    fraction = pos - int(pos)
    return data[lower] + fraction * (data[upper] - data[lower])

Q1 = get_quartile(clean_mdn_house, 1)
Q3 = get_quartile(clean_mdn_house, 3)
print(f"The value of Q1={Q1} and Q3={Q3}")

plt.figure(figsize=(10,5))
plt.boxplot(clean_mdn_house,
            vert=True,
            patch_artist=True,
            notch=True)
plt.axhline(Q1, color="red", linestyle="--", label=f"Lower: {Q1:.2f}")
plt.axhline(Q3, color="green", linestyle=":", label=f"Upper: {Q3:.2f}")
plt.legend()
plt.show()
```

**12. Plot density plots for two different columns on the same figure and analyze overlap.**

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Clean data
clean_mdn_house = [x for x in mdn_house if x is not None]
clean_mdn_income = [x for x in mdn_income if x is not None]

# Min-max normalization (pure Python)
def normalize(data):
    min_val = min(data)
    max_val = max(data)
    return [(x - min_val) / (max_val - min_val) for x in data]

house_norm = normalize(clean_mdn_house)
income_norm = normalize(clean_mdn_income)

# Plot both density curves on same figure
plt.figure(figsize=(12,6))
sns.kdeplot(house_norm, fill=True, label="Median House Value (Normalized)")
sns.kdeplot(income_norm, fill=True, label="Median Income (Normalized)")

plt.title("Density Plot Comparison of Two Columns")
plt.xlabel("Normalized Value")
plt.ylabel("Density")
plt.legend()
plt.show()
```

The overlapping area of the two curves indicates ranges where both variables share similar relative distributions, suggesting a potential relationship between income levels and house values. Normalization was applied to bring both columns to a common scale (0–1), enabling a fair comparison of their distribution shapes without distortion caused by differing units and magnitudes.**

**13. Explain how skewness is visible in histograms and boxplots.**

In a **histogram**, skewness is identified by the length of the tail. If the bars extend longer toward the right side (higher values), the distribution is positively skewed (right-skewed), meaning most values are concentrated on the lower side with a few large values pulling the tail right. If the tail extends toward the left side (lower values), the distribution is negatively skewed (left-skewed). A symmetric distribution shows roughly equal spread on both sides of the center.

In a **boxplot**, skewness is seen through the position of the median, the length of the whiskers, and the location of outliers. For positive skewness, the median is closer to Q1, the upper whisker is longer, and outliers often appear on the higher-value side. For negative skewness, the median is closer to Q3, the lower whisker is longer, and outliers appear on the lower side. If the median is centered and whiskers are similar in length, the distribution is approximately symmetric.

```python
def get_quantiles(data,p):
    if not data:
        return None
    
    data = sorted(data)
    n = len(data)
    pos = (p/100) * (n-1)

    lower = int(pos) 
    upper = lower+1

    if upper>=n:
        return data[lower]
        
    fraction = pos - int(pos)
    return data[lower] + fraction * (data[upper] - data[lower])

clean_mdn_house = [x for x in mdn_house if x is not None]
ten_per= get_quantiles(clean_mdn_house,10)
tnyfive_per= get_quantiles(clean_mdn_house,25)
fifty_per= get_quantiles(clean_mdn_house,50)
svntyfive_per= get_quantiles(clean_mdn_house,75)
ninety_per= get_quantiles(clean_mdn_house,90)
print(f"The value in 10% quantiles is {ten_per}")
print(f"The value in 25% quantiles is {tnyfive_per}")
print(f"The value in 50% quantiles is {fifty_per}")
print(f"The value in 75% quantiles is {svntyfive_per}")
print(f"The value in 90% quantiles is {ninety_per}")


```

**Quantiles divide a dataset into equal parts based on percentages. The 10%, 25%, 50%, 75%, and 90% quantiles represent the values below which 10%, 25%, 50%, 75%, and 90% of the data lie, respectively.
In a boxplot, the 25% quantile (Q1) and 75% quantile (Q3) form the lower and upper edges of the box, while the 50% quantile (median) is shown as a line inside the box. The interquartile range (IQR = Q3 − Q1) measures data spread. Whiskers extend to the smallest and largest values within 1.5 × IQR, and values beyond this range are plotted as outliers.
The 10% and 90% quantiles help understand the distribution tails but are not used directly in boxplot construction or outlier detection**

**24. Write a function that flags extreme outliers (beyond 3 × IQR) and visualize them.**

```python
import matplotlib.pyplot as plt

def get_quartile(data, q):
    """Returns the q-th quartile of a numeric dataset."""
    if not data:
        return None
    if q not in (1, 2, 3):
        print("Error: Quartiles can only be 1, 2, or 3.")
        return None
    
    data = sorted(data)
    n = len(data)
    pos = q * (n + 1) / 4

    if pos <= 1:
        return data[0]
    elif pos >= n:
        return data[-1]
    
    lower = int(pos) - 1
    upper = lower + 1
    fraction = pos - int(pos)
    return data[lower] + fraction * (data[upper] - data[lower])


def flag_extreme_outliers(data):
    """Flags extreme outliers (beyond 3 × IQR) and returns a dictionary with flags."""
    clean_data = [x for x in data if x is not None]
    
    q1 = get_quartile(clean_data, 1)
    q3 = get_quartile(clean_data, 3)
    iqr = q3 - q1
    
    lower_bound = q1 - 3 * iqr
    upper_bound = q3 + 3 * iqr
    
    flags = ["Extreme Outlier" if (x < lower_bound or x > upper_bound) else "Normal" for x in clean_data]
    
    return clean_data, flags, lower_bound, upper_bound


# Apply the function
clean_data, flags, lower, upper = flag_extreme_outliers(mdn_income)

# Print summary
print("Total number of valid data points:", len(clean_data))
print("Number of extreme outliers detected:", flags.count("Extreme Outlier"))

# Visualize using a boxplot
plt.figure(figsize=(10, 5))
plt.boxplot(clean_data, vert=True, patch_artist=True,)
plt.title("Boxplot of Data with Extreme Outliers Flagged")
plt.xlabel("Values")
plt.show()
```

**25. Compare IQR-based outliers vs Z-score-based outliers and visualize both.**

```python
import matplotlib.pyplot as plt

# Function to calculate mean
def mean(data):
    if not data:
        return None
    return sum(data) / len(data)

# Function to calculate standard deviation
def std_dev(data):
    m = mean(data)
    var = sum((x - m) ** 2 for x in data) / len(data)
    return var ** 0.5

# Function to get quartiles
def get_quartile(data, q):
    """Returns the q-th quartile of a numeric dataset."""
    if not data:
        return None
    if q not in (1, 2, 3):
        print("Error: Quartiles can only be 1, 2, or 3.")
        return None
    
    data = sorted(data)
    n = len(data)
    pos = q * (n + 1) / 4

    if pos <= 1:
        return data[0]
    elif pos >= n:
        return data[-1]
    
    lower = int(pos) - 1
    upper = lower + 1
    fraction = pos - int(pos)
    return data[lower] + fraction * (data[upper] - data[lower])

# Function to get outliers using IQR or Z-score
def get_outliers(data, method="iqr"):
    clean_data = [x for x in data if x is not None]
    outliers = []

    if method == "iqr":
        q1 = get_quartile(clean_data, 1)
        q3 = get_quartile(clean_data, 3)
        iqr = q3 - q1
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr
        outliers = [x for x in clean_data if x < lower_bound or x > upper_bound]

    elif method == "zscore":
        m = mean(clean_data)
        std = std_dev(clean_data)
        outliers = [x for x in clean_data if abs((x - m) / std) > 3]

    return outliers

# Clean data (exclude None)
clean_data = [x for x in mdn_income if x is not None]

# Get outliers
iqr_outliers = get_outliers(clean_data, method="iqr")
zscore_outliers = get_outliers(clean_data, method="zscore")

# Print summary
print("Number of IQR outliers:", len(iqr_outliers))
print("Number of Z-score outliers:", len(zscore_outliers))

# Visualization
plt.figure(figsize=(12, 6))

# Boxplot for IQR method (red outliers)
plt.boxplot(clean_data, vert=False, patch_artist=True,
            flierprops=dict(marker='x', color='red', markersize=8, label='IQR Outliers'))

# Overlay Z-score outliers (blue)
first =True
for i, x in enumerate(clean_data):
    if x in zscore_outliers:
        plt.scatter(x, 1, color='blue', s=50,
                    label='Z-score Outlier' if first else "")
        first = False


plt.title("Comparison of Outliers: IQR (red) vs Z-score (blue)")
plt.xlabel("Values")
plt.yticks([1], ["Data"])
plt.legend()
plt.show()
```

**30. Identify which visualization is most robust to outliers and justify your answer.**

The box plot is the most robust visualization to outliers because it uses the median and interquartile range (IQR) to summarize central tendency and spread, while plotting outliers as separate points. This ensures that extreme values do not distort the main visual representation of the data.
Histograms rely on bin counts across the full data range, so outliers can stretch the axis and concentrate most observations into a few bins, hiding the true distribution of the majority of the data. Kernel Density Estimation (KDE) uses all data points to estimate a smooth density curve; extreme values influence the curve shape and bandwidth, often introducing long tails or skew that misrepresents the central distribution.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# ---------- Custom Statistics ----------
def mean(data):
    return sum(data) / len(data)

def median(data):
    data = sorted(data)
    n = len(data)
    mid = n // 2
    if n % 2 == 0:
        return (data[mid - 1] + data[mid]) / 2
    else:
        return data[mid]

def std_dev(data):
    m = mean(data)
    var = sum((x - m) ** 2 for x in data) / len(data)
    return var ** 0.5

# ---------- Quartile / Outlier Functions ----------
def get_quartile(data, q):
    if not data:
        return None
    if q not in (1,2,3):
        print("Quartiles can only be 1,2,3")
        return
    data = sorted(data)
    n = len(data)
    pos = q * (n + 1) / 4
    if pos <= 1:
        return data[0]
    if pos >= n:
        return data[-1]
    lower = int(pos) - 1
    upper = lower + 1
    fraction = pos - int(pos)
    return data[lower] + fraction * (data[upper] - data[lower])

def remove_outliers(data):
    data = [x for x in data if x is not None]
    if not data:
        return []
    q1 = get_quartile(data, 1)
    q3 = get_quartile(data, 3)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return [x for x in data if lower <= x <= upper]

# ---------- Visualization Pipeline ----------
def visualization_pipeline(raw_data, feature_name, color="skyblue"):
    # Step 1: Remove outliers
    clean_data = remove_outliers(raw_data)
    if not clean_data:
        print(f"No valid data for {feature_name}")
        return
    
    # Step 2: Compute statistics
    m = mean(clean_data)
    med = median(clean_data)
    s = std_dev(clean_data)
    
    # Print stats
    print(f"{feature_name} → Mean: {m:.2f}, Median: {med:.2f}, Std: {s:.2f}")
    
    # Step 3: Plot KDE
    plt.figure(figsize=(6,4))
    sns.kdeplot(clean_data, fill=True, color=color)
    plt.axvline(m, color="red", linestyle="--", label=f"Mean: {m:.2f}")
    plt.axvline(med, color="green", linestyle=":", label=f"Median: {med:.2f}")
    plt.axvline(s, color="orange", linestyle="-.", label=f"Std: {s:.2f}")
    plt.title(f"KDE of {feature_name}")
    plt.xlabel(feature_name)
    plt.ylabel("Density")
    plt.legend()
    plt.show()

# ---------- Example Usage ----------
clean_mdn_housemdn_house = [x for x in mdn_house if x is not None]
clean_mdn_income = [x for x in mdn_income if x is not None]

visualization_pipeline(clean_mdn_house, "Median House", color="skyblue")
visualization_pipeline(clean_mdn_income, "Median Income", color="lightgreen")
```
